```python
import numpy as np
```

```python
class GDRegressor:

    def __init__(self, learning_rate = 0.01, epochs = 1000):
        self.coef_ = None
        self.intercept_ = None
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.loss_history = []
        
    def fit(self, X_train, y_train):

        self.coef_ = 0
        self.intercept_ = 0

        for i in range(self.epochs):
            y_pred = self.coef_ * X_train.ravel() + self.intercept_
            error = y_train - y_pred
            loss_function = np.mean(error ** 2)
            
            gradient_coef = -2 * np.mean(error * X_train.ravel())
            gradient_intercept  = -2 * np.mean(error)

            self.coef_ -= self.learning_rate * gradient_coef
            self.intercept_ -= self.learning_rate * gradient_intercept

            self.loss_history.append(loss_function)

            if (i + 1) %100 == 0 or i== 0:
                print(f"Epoch {i+1}/{self.epochs} - MSE:{loss_function: .3f} - Coef:{self.coef_: .3f}- Intercept: {self.intercept_: .3f}")


    def predict(self, X_test):

        return self.coef_ * X_test + self.intercept_



```
